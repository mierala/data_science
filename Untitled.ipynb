# 1. Business Understanding

### Latar Belakang Masalah
Sebuah perusahaan retail mengalami pemborosan anggaran promosi karena menyamaratakan strategi pemasaran ke semua pelanggan, sehingga mereka ingin mengelompokkan ribuan data profil konsumen berdasarkan pendapatan (`Income`), pengeluaran produk premium (`MntWines`), dan frekuensi belanja promo (`NumDealsPurchases`) guna menemukan kelompok pelanggan dengan karakteristik serupa agar penawaran promosi ke depannya dapat dirancang lebih hemat biaya dan tepat sasaran.

### Rumusan Masalah
1. Strategi promosi yang disamaratakan (*one-size-fits-all*) menimbulkan inefisiensi anggaran pemasaran yang signifikan.
2. Konsumen berpenghasilan tinggi (*affluent*) yang bersedia membayar harga normal justru sering diberi diskon besar yang menggerus profit margin (*margin leakage*).
3. Konsumen beranggaran terbatas (*budget-conscious*) sering ditawari produk mewah (*premium wines*) yang tidak terjangkau sehingga tingkat konversi penawaran menjadi sangat rendah.
4. Perusahaan belum memiliki segmentasi pelanggan berbasis data (*data-driven customer segmentation*) untuk membedakan kelompok pemburu promo, konsumen setia produk premium, dan konsumen hemat.

## Alasan Memilih Dataset
Dataset `marketing_campaign.csv` dipilih dengan pertimbangan utama:
1. **Relevansi Fitur Finansial dan Perilaku**: Dataset ini mencatat data profil demografis (`Income`), riwayat pengeluaran kategori produk premium (`MntWines`), serta rekam jejak transaksi yang memanfaatkan promosi diskon (`NumDealsPurchases`). Ketiga variabel ini merupakan pilar esensial untuk membedakan konsumen bernilai tinggi (*high-value*), pemburu diskon (*deal seekers*), dan konsumen hemat (*budget-conscious*).
2. **Kesesuaian dengan Permasalahan Retail Nyata**: Dataset merefleksikan dinamika interaksi pelanggan dengan berbagai kampanye promosi retail yang nyata.
3. **Kualitas dan Kelayakan Data**: Memiliki ukuran sampel yang memadai (2.240 observasi) sehingga sangat ideal untuk penerapan algoritma *unsupervised learning* tanpa risiko *underfitting*.

### Tujuan Bisnis (Business Objectives)
- Mengidentifikasi kelompok profil pelanggan yang homogen guna menghindari pemborosan anggaran promosi.
- Merancang rekomendasi strategi promosi yang spesifik, terukur, dan terpersonalisasi untuk setiap segmen pelanggan (*cost-effective targeted marketing*).

### Sasaran Data Mining (Data Mining Goals)
- Melakukan segmentasi pelanggan menggunakan algoritma clustering (*K-Means Clustering*).
- Menentukan jumlah klaster optimal ($k$) menggunakan *Elbow Method* dan *Silhouette Analysis*.
- Mengevaluasi keandalan dan separasi klaster menggunakan metrik *Silhouette Score*, *Davies-Bouldin Index*, dan *Calinski-Harabasz Index*.
- Membangun model deployment berupa fungsi inferensi untuk memprediksi klaster profil pelanggan baru secara otomatis.

# 3. Data Preparation

Pada tahap Data Preparation, dilakukan rangkaian pra-pemrosesan data agar siap digunakan oleh algoritma clustering:
1. **Handling Missing Values**: Menghilangkan baris dengan nilai `Income` kosong.
2. **Filtering Outliers**: Menghilangkan data pelanggan dengan nilai `Income` anomali (> $120.000).
3. **Feature Selection**: Memilih ketiga fitur fokus (`Income`, `MntWines`, `NumDealsPurchases`).
4. **Feature Scaling (Standardisasi)**: Melakukan penskalaan data menggunakan `StandardScaler` agar seluruh fitur memiliki nilai mean = 0 dan standar deviasi = 1, mencegah dominasi fitur berdimensi besar terhadap perhitungan jarak Euclidean.

```python
# 1. Menghapus missing values pada Income
df_clean = df.dropna(subset=['Income']).copy()

# 2. Menghilangkan outlier ekstrem Income (> $120.000)
df_clean = df_clean[df_clean['Income'] < 120000].copy()

print(f"Jumlah baris awal              : {df.shape[0]}")
print(f"Jumlah baris setelah dibersihkan: {df_clean.shape[0]}")
print(f"Total baris yang difilter       : {df.shape[0] - df_clean.shape[0]}")
```

```python
# 3. Pemilihan Fitur
features = ['Income', 'MntWines', 'NumDealsPurchases']
X = df_clean[features].copy()

# 4. Standardisasi Data menggunakan StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Konversi kembali ke DataFrame untuk kemudahan inspeksi
X_scaled_df = pd.DataFrame(X_scaled, columns=features, index=df_clean.index)
print("5 Baris Teratas Data Setelah Standarisasi:")
X_scaled_df.head()
```

```python
# Verifikasi statistik data terstandarisasi (Mean mendekati 0, Std mendekati 1)
print("Statistik Data Setelah Standarisasi:")
X_scaled_df.describe().round(3)
```

# 4. Modeling

Tahapan pemodelan mencakup:
1. **Menentukan Jumlah Cluster Optimal ($k$)**:
   - **Elbow Method (Inertia / WCSS)**: Menilai titik pembelokan inersia.
   - **Silhouette Score Analysis**: Menilai kualitas pemisahan klaster untuk nilai $k \in [2, 10]$.
2. **Membangun Model K-Means Clustering** menggunakan nilai $k$ terbaik.
3. **Visualisasi Klaster**: Memetakan sebaran pelanggan dalam ruang 2D dan 3D.

```python
# Perhitungan Inersia (Elbow Method) dan Silhouette Score untuk rentang k = 2 s/d 10
inertia = []
silhouette_scores = []
k_range = range(2, 11)

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertia.append(km.inertia_)
    silhouette_scores.append(silhouette_score(X_scaled, labels))
```

```python
# Visualisasi Elbow Method dan Silhouette Score
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# Plot Elbow
axes[0].plot(k_range, inertia, marker='o', color='#1f77b4', linewidth=2, markersize=7)
axes[0].set_title('Elbow Method (Penurunan Inertia/WCSS)', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Jumlah Cluster (k)', fontsize=11)
axes[0].set_ylabel('Inertia (WCSS)', fontsize=11)
axes[0].set_xticks(k_range)
axes[0].grid(True, linestyle='--', alpha=0.6)

# Plot Silhouette
axes[1].plot(k_range, silhouette_scores, marker='s', color='#ff7f0e', linewidth=2, markersize=7)
axes[1].axvline(x=3, color='red', linestyle='--', label='k optimal = 3')
axes[1].set_title('Silhouette Score per Nilai k', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Jumlah Cluster (k)', fontsize=11)
axes[1].set_ylabel('Silhouette Score', fontsize=11)
axes[1].set_xticks(k_range)
axes[1].legend()
axes[1].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()

best_k = k_range[np.argmax(silhouette_scores)]
print(f"Konfigurasi optimal terpilih: k = {best_k} dengan Silhouette Score = {max(silhouette_scores):.4f}")
```

### Keputusan Jumlah Cluster
Berdasarkan pengujian komparatif di atas:
- **Elbow Method**: Penurunan drastis inersia mulai melandai secara bertahap setelah $k=3$.
- **Silhouette Score**: Mencapai puncak nilai tertinggi pada $k=3$ sebesar **0.4710**. Nilai ini menunjukkan bahwa pembagian ke dalam 3 kelompok menghasilkan separasi yang paling tegas dan kohesi internal yang paling solid.

Maka, model K-Means akan dibangun menggunakan **$k = 3$**.

```python
# Menerapkan K-Means Clustering dengan k = 3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_clean['Cluster'] = kmeans.fit_predict(X_scaled)

# Menampilkan distribusi jumlah anggota tiap klaster
print("Distribusi Pelanggan pada Setiap Cluster:")
cluster_counts = df_clean['Cluster'].value_counts().sort_index()
for c_id, count in cluster_counts.items():
    pct = (count / len(df_clean)) * 100
    print(f"Cluster {c_id}: {count:,} pelanggan ({pct:.2f}%)")
```

```python
# Visualisasi Hasil Segmentasi 2 Dimensi
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
cluster_colors = {0: '#4daf4a', 1: '#377eb8', 2: '#e41a1c'}

# Scatter 1: Income vs MntWines
sns.scatterplot(
    data=df_clean, x='Income', y='MntWines', hue='Cluster',
    palette=cluster_colors, alpha=0.65, s=60, ax=axes[0]
)
axes[0].set_title('Hasil Segmentasi: Income vs MntWines', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Pendapatan (Income)', fontsize=11)
axes[0].set_ylabel('Pengeluaran Wine (MntWines)', fontsize=11)

# Scatter 2: Income vs NumDealsPurchases
sns.scatterplot(
    data=df_clean, x='Income', y='NumDealsPurchases', hue='Cluster',
    palette=cluster_colors, alpha=0.65, s=60, ax=axes[1]
)
axes[1].set_title('Hasil Segmentasi: Income vs NumDealsPurchases', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Pendapatan (Income)', fontsize=11)
axes[1].set_ylabel('Frekuensi Pembelian Promo (NumDealsPurchases)', fontsize=11)

plt.tight_layout()
plt.show()
```

```python
# Visualisasi 3D Ruang Klaster (Income x MntWines x NumDealsPurchases)
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

for c_id in sorted(df_clean['Cluster'].unique()):
    sub = df_clean[df_clean['Cluster'] == c_id]
    ax.scatter(
        sub['Income'], sub['MntWines'], sub['NumDealsPurchases'],
        c=cluster_colors[c_id], label=f'Cluster {c_id}', alpha=0.5, s=35
    )

ax.set_xlabel('Income ($)', labelpad=12, fontweight='bold')
ax.set_ylabel('MntWines ($)', labelpad=12, fontweight='bold')
ax.set_zlabel('NumDealsPurchases', labelpad=10, fontweight='bold')
ax.set_title('Visualisasi Ruang 3D Customer Clusters', fontsize=14, fontweight='bold')
ax.legend(title='Klaster', loc='upper left')
plt.tight_layout()
plt.show()
```

# 5. Evaluation

Tahap evaluasi bertujuan untuk mengukur performa teknis dari model pengelompokan tanpa supervisi (*unsupervised learning*) menggunakan tiga metrik standar industri:
1. **Silhouette Score**: Mengukur kohesi dalam klaster vs jarak ke klaster terdekat. Rentang [-1, 1], semakin tinggi semakin baik.
2. **Davies-Bouldin Index (DBI)**: Mengukur rasio kemiripan antar-klaster. Nilai semakin mendekati 0 mengindikasikan klaster yang lebih terpisah secara tegas.
3. **Calinski-Harabasz Index (CHI)**: Rasio antara dispersi antar-klaster dan dispersi intra-klaster. Semakin tinggi nilainya, semakin baik partisi yang dihasilkan.

Untuk memperkuat validasi, performa K-Means dikomparasikan dengan algoritma clustering alternatif: **Hierarchical Clustering (Agglomerative)**.

```python
# Melatih model alternatif untuk perbandingan evaluasi: Hierarchical (Agglomerative) Clustering
agg = AgglomerativeClustering(n_clusters=3)
agg_labels = agg.fit_predict(X_scaled)

# Menghitung metrik evaluasi
eval_data = [
    {
        'Algoritma': 'K-Means (k=3)',
        'Silhouette Score': silhouette_score(X_scaled, df_clean['Cluster']),
        'Davies-Bouldin Index': davies_bouldin_score(X_scaled, df_clean['Cluster']),
        'Calinski-Harabasz Index': calinski_harabasz_score(X_scaled, df_clean['Cluster'])
    },
    {
        'Algoritma': 'Hierarchical Clustering (k=3)',
        'Silhouette Score': silhouette_score(X_scaled, agg_labels),
        'Davies-Bouldin Index': davies_bouldin_score(X_scaled, agg_labels),
        'Calinski-Harabasz Index': calinski_harabasz_score(X_scaled, agg_labels)
    }
]

eval_df = pd.DataFrame(eval_data).round(4)
print("=== Tabel Perbandingan Hasil Evaluasi Model Clustering ===")
eval_df
```

### Analisis dan Kesimpulan Evaluasi
1. **Keunggulan K-Means**: K-Means ($k=3$) menghasilkan performa terbaik secara konsisten pada semua metrik evaluasi, dengan **Silhouette Score tertinggi (0.4710)**, **Davies-Bouldin Index terendah (0.8301)**, dan **Calinski-Harabasz Index tertinggi (2076.92)** dibandingkan Hierarchical Clustering.
2. **Kualitas Klaster**: Nilai DBI yang rendah (< 1.0) dan skor siluet yang solid memastikan bahwa masing-masing kelompok konsumen terpisah dengan batas yang jelas (*well-separated*) serta memiliki homogenitas internal yang baik (*compact*).

```python
# Fungsi Inferensi untuk Deployment: Prediksi Segmen Pelanggan Baru
def predict_customer_segment(income, mnt_wines, num_deals):
    """
    Fungsi untuk memprediksi klaster dan memberikan rekomendasi promosi
    bagi profil pelanggan retail baru.
    """
    # Memuat kembali model dan scaler
    loaded_scaler = joblib.load('scaler_customer_segmentation.pkl')
    loaded_model = joblib.load('kmeans_customer_segmentation.pkl')
    
    # Transformasi data masukan
    input_data = np.array([[income, mnt_wines, num_deals]])
    input_scaled = loaded_scaler.transform(input_data)
    
    # Prediksi klaster
    cluster_pred = int(loaded_model.predict(input_scaled)[0])
    
    # Metadata profil dan aksi bisnis
    strategy_mapping = {
        0: {
            "segmen": "Deal Seekers (Pemburu Promo)",
            "profil": "Pendapatan menengah, konsumsi wine sedang, sangat aktif berbelanja dengan promo.",
            "rekomendasi": "Kirimkan penawaran diskon, voucher belanja waktu terbatas, dan promo bundle hemat."
        },
        1: {
            "segmen": "Budget-Conscious (Pelanggan Hemat)",
            "profil": "Pendapatan rendah, pengeluaran wine minim, sensitif terhadap harga barang pokok.",
            "rekomendasi": "Jangan promosikan produk premium. Fokus pada produk esensial terjangkau dan gratis ongkir."
        },
        2: {
            "segmen": "Affluent Connoisseurs (Konsumen Premium/Makmur)",
            "profil": "Pendapatan tinggi, pengeluaran wine sangat tinggi, tidak terdorong oleh diskon.",
            "rekomendasi": "Hentikan pemberian diskon promo. Berikan layanan VIP, tasting event eksklusif, dan produk limited edition."
        }
    }
    
    return {
        "Cluster": cluster_pred,
        "Nama Segmen": strategy_mapping[cluster_pred]["segmen"],
        "Profil Singkat": strategy_mapping[cluster_pred]["profil"],
        "Rekomendasi Strategi": strategy_mapping[cluster_pred]["rekomendasi"]
    }

# Uji Coba Deployment dengan Tiga Skenario Pelanggan Baru
simulasi_pelanggan = [
    {"nama": "Pelanggan A (Budi)", "income": 53000, "mnt_wines": 340, "num_deals": 8},
    {"nama": "Pelanggan B (Siti)", "income": 28000, "mnt_wines": 30,  "num_deals": 1},
    {"nama": "Pelanggan C (Darma)", "income": 82000, "mnt_wines": 750, "num_deals": 1}
]

print("=== Simulasi Prediksi dan Rekomendasi Sistem Deployment ===")
for p in simulasi_pelanggan:
    hasil = predict_customer_segment(p["income"], p["mnt_wines"], p["num_deals"])
    print(f"\n{p['nama']} [Income: ${p['income']:,}, MntWines: ${p['mnt_wines']}, Deals: {p['num_deals']}]:")
    print(f"  -> Klaster Terprediksi: Cluster {hasil['Cluster']} ({hasil['Nama Segmen']})")
    print(f"  -> Profil             : {hasil['Profil Singkat']}")
    print(f"  -> Rekomendasi Promosi: {hasil['Rekomendasi Strategi']}")
```
